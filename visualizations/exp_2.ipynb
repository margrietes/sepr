```python
import seaborn as sns
import pandas as pd

from pathlib import Path; file_path = Path.cwd().parents[1] / "code" / "results" / "exp_2_sample.csv"
data = pd.read_csv(file_path)
```

Do pC outcomes increase relatively to pD outcomes with increasing p?

```python
# Frequencies of pC > pD = 1
sns.stripplot(data=data, x="p", y="pC", alpha=0.5, size=5)
sns.pointplot(data=data, x="p", y="pC", errorbar="ci")
```

```python
# Density of pC values of different p values
sns.kdeplot(data=data, x="pC", hue="p", fill=True, alpha=0.1)
```
