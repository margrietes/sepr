```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path; file_path = Path.cwd().parents[1] / "code" / "results" / "exp_2_full.csv"
data = pd.read_csv(file_path)
```

```python
data
```

Do pC outcomes increase relatively to pD outcomes with increasing p?

```python
# Boxplot of p values for each strategy
ax = sns.boxplot(data=data, x='pC > pD', y='p')
ax.set_ylabel("Probability of transitioning to another network state")
```

```python
# Boxplot of pC values for each p
fig, ax = plt.subplots(1,2, figsize=(8, 6), sharex=True)
sns.boxplot(data=data, x='p', y='pC', width=0.6, ax=ax[0])
sns.boxplot(data=data, x='p', y='pD', width=0.6, ax=ax[1])
plt.xlabel("Probability of transitioning to another network state")
ax[0].set_ylabel("Probability of Cooperation (pC)")
ax[1].set_ylabel("Probability of Defection (pD)")
```

```python
# Line plot of pC values and pD values for each p
ax = sns.lineplot(data=data, x="p", y="pC", label="pC", color="green", marker="o")
sns.lineplot(data=data, x="p", y="pD", label="pD", color="red", marker="s", ax=ax)
ax.set_ylim(0, 0.25)
ax.set_xlabel("Probability of transitioning to another network state")
ax.set_ylabel("Probability of Cooperation (pC) or Defection (pD)")
```

```python
# Density of pC outcomes and pD outcomes for each p value
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.kdeplot(data=data, x="pC", hue="p", fill=True, alpha=0.3, ax=ax[0])
sns.kdeplot(data=data, x="pD", hue="p", fill=True, alpha=0.3, ax=ax[1], legend=False)
```
